### 01: Get data from config file via get_api_info function

```python
from src.utils.get_api_info import get_api_info

api_key_secret, api_scope_secret, base_url = get_api_info()

api_key = dbutils.secrets.get(scope=api_scope_secret, key=api_key_secret)
```

### 02: Call the market_api function

```python
from src.api.market_data import get_market_data

# Fetch market data (limit pages during testing to avoid rate limits)
market_data = get_market_data(api_key, base_url, 250) # 250 = Max Pages

print(f"Retrieved {len(market_data)} coins")
if market_data:
    print(f"\nSample data:")
    print(f"  {market_data[0]['id']}: ${market_data[0]['current_price']}")
    print(f"  {market_data[-1]['id']}: ${market_data[-1]['current_price']}")
```

### 03: save the raw JSON in Volumes

```python
from src.raw.raw_storage import save_raw_json

save_raw_json(
    data=market_data,
    path="/Volumes/coin_gecko_dev/raw/api_files/",
    file_type="market",
)
```
